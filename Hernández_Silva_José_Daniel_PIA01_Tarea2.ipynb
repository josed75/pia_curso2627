#Ejercicio 8
Muestra los datos de la fila 3 del DataFrame. A continuación, muestra la media de edad de las personas cuyos datos están almacenados en el DataFrame y la proporción de mujeres existentes con respecto al total de personas.

```python
print(seguro.iloc[3])

print("Media de edad:", seguro["age"].mean())

proporcion_mujeres = (seguro["sex"] == "female").sum() / len(seguro)
print("Proporción de mujeres:", proporcion_mujeres)
```

#Ejercicio 9
Crea una copia del DataFrame e inserta en esta copia una nueva fila con unos datos ficticios. Muestra por pantalla los tres últimos registros del DataFrame original y los tres últimos registros de la copia.

```python
copia = seguro.copy()

nueva_fila = {
    "age": 25,
    "sex": "female",
    "bmi": 25.0,
    "children": 0,
    "smoker": "no",
    "region": "southwest",
    "charges": 3000.0
}

copia.loc[len(copia)] = nueva_fila

print("Últimos 3 registros del DataFrame original:")
print(seguro.tail(3))

print("Últimos 3 registros de la copia:")
print(copia.tail(3))
```
